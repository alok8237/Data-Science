Deep and Shallow copy

```python
a=np.arange(1,21)
a
```

```python
slice=a[:5]
slice=slice*10
slice
```

```python
a[4:6]*10
```

```python
a
b=a  #deep copy
b[0]=99
b
```

```python
a
```

```python
c=np.arange(1,21)
c
```

```python
d=c.copy() #shallow copy
d[0]=88
d
```

Matrix operations

```python
A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])
A,B
```

```python
A@B #matrix multiplication
```

```python
A*B
```

```python
np.dot(A,B) #matrix multiplication
```

```python
A.T #transpose
```

**Advance Array Manipulation**

Stacking Array

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
```

```python
np.vstack((a,b)) #vertical stack
```

```python
np.hstack((a,b)) #horizontal stack
```

```python
np.column_stack((a,b))
```

Splitting array

```python
c=np.arange(16).reshape(4,4)
c
```

```python
np.hsplit(c,2)
```

```python
np.hsplit(c,4) #horizontal split
```

```python
np.vsplit(c,2) #vertical split
```

```python
 a=np.vsplit(c,4)
```

```python
for i in a:
    print(i)
```
