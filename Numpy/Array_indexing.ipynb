**numpy indexing and slicing of vectors**

```python
arr=np.arange(11,21)
arr
```

```python
arr[6] #indexing
```

```python
arr[1:5] #slicing
```

```python
arr[:5]
```

```python
arr[3:]
```

```python
arr[:]
```

```python
arr[3::2]
```

**numpy indexing and slicing of matrix**

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[1]
```

```python
arr[0,0] #0th row and 0th column
```

```python
arr[0,4]
```

```python
arr[5][4]
```

```python
slice=arr[0:2,1:3]
slice
```

```python
slice=arr[3:,3:]
slice
```

```python
arr[:,2]
```

**Boolean indexing**

```python
arr=np.arange(11,21)
arr
```

```python
bool_index=arr%2==0
bool_index
```

```python
arr=arr[bool_index]
arr
```
