**Valid Sudoku**

```python
s=np.array([
    [5,3,4,6,7,8,9,1,2],
    [6,7,2,1,9,5,3,4,8],
    [1,9,8,3,4,2,5,6,7],

    [8,5,9,7,6,1,4,2,3],
    [4,2,6,8,5,3,7,9,1],
    [7,1,3,9,2,4,8,5,6],

    [9,6,1,5,3,7,2,8,4],
    [2,8,7,4,1,9,6,3,5],
    [3,4,5,2,8,6,1,7,9]
])
```

```python
b=np.sum(s,axis=1)
for i in b:
    if i!=45:
        print("Sudoko is not valid")
        break
else:
    print("For rows it is valid")
```

```python
c=np.sum(s,axis=0)
for i in c:
    if i!=45:
        print("Sudoko is not valid")
        break
else:
    print("For columns it is valid")
```

```python
s1=0
for i in range(0,9,3):
    for j in range(0,9,3):
        d=np.sum(s[i:i+3,j:j+3])
        print(d)
        s1+=1
if s1==9:
    print("Valid Sudoko")
else:
    print("Sudoko is not valid")
```

```python
#Columns: [Age,Maths marks,Science marks]
data=np.array([
    [18,85,78],
    [19,92,88],
    [17,76,95],
    [18,65,70],
    [20,90,85]
])
```

```python
#Find the average age of students
sum=0
for i in range(0,data.shape[0]):
    sum+=data[i][0]
avg=sum/data.shape[0]
print(f"Average of age is {avg}")
    
```
